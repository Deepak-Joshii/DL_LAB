# MLP - Multi Layer Perceptron

### import libraries 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Define Activation Function

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

### Derivative of back Propagation

```python
def derivative_sigmoid(z):
    s = sigmoid(z)
    return s * (1 - s)
```

### Initialize X and Y

```python
X=np.array([1,2,3,4,5])
y=2*X
```

### Intialise Weights ans Biases

```python
w_h=np.random.rand(1)[0]
b_h=np.random.rand(1)[0]
w_o=np.random.rand(1)[0]
b_o=np.random.rand(1)[0]    
```

```python
epoch=50
learning_rate=0.01
losses=[]
```

```python
for i in range(epoch):
    #forward_pass
    z1=X*w_h+b_h
    a1=sigmoid(z1)
    
    z2=w_o*a1+b_o
    y_pred=sigmoid(z2)
    #Loss(MSE)
    loss=np.mean((y-y_pred)**2)
    print(f"Epoch{i+1},Losss:{loss}")
    losses.append(loss)

    #Backward pass

    error=y_pred-y
    err_o=error*derivative_sigmoid(z2)

    hid_err=err_o*w_o
    err_hid=hid_err*derivative_sigmoid(z1)

    #Update weights and biases
    w_o=w_o+learning_rate*(y-y_pred)*a1
    b_o=b_o+learning_rate*(y-y_pred)

    w_h=w_h+learning_rate*(y-y_pred)*X
    b_h=b_h+learning_rate*(y-y_pred)


    
    
```
